```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)

df = iris.frame
```

```python
X = df.iloc[:, :-1].to_numpy()
y = df.iloc[:, -1].to_numpy()
```

```python
def inhu_shupple(X:np.ndarray, y:np.ndarray):
    rand_idx = np.random.permutation(len(X))
    X,y = X[rand_idx], y[rand_idx]
    return X,y

def inhu_standard_scaler(X:np.ndarray):
    return (X - X.mean())/X.std()
    
def inhu_train_test_split(X:np.ndarray, y:np.ndarray, ratio:float = 0.8):
    threshold = int(len(X) * ratio)
    X_train, X_test, y_train, y_test = X[:threshold], X[threshold:], y[:threshold], y[threshold:]
    return X_train, X_test, y_train, y_test

def inhu_soft_max(prob:np.ndarray) -> np.ndarray:
    exp_list = np.exp(prob)
    return exp_list/np.sum(exp_list)

#todo : turn signoid to MLE then calc the minimum using mini batch
```
