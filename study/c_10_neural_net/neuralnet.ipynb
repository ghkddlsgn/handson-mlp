```python
learning_rate = 0.1
x = torch.tensor(5.0, requires_grad=True)
for i in range(100):
    f = x ** 2
    f.backward()
    with torch.no_grad():
        x -= learning_rate * x.grad
    
    x.grad.zero_()
```
